```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import string
```

```python
nltk.download("punkt")
nltk.download("stopwords")
```

```python
text="I absolutely loved this movie!"
text=text.lower()
print(text)
text=text.translate(str.maketrans('','',string.punctuation))
print(text)
```

```python
token=word_tokenize(text)
print(token)
stop_word=set(stopwords.words('english'))
filterwords=[word for word in token if word not in stop_word]
print(filterwords)
```

```python
stemmer=PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filterwords]
print(stemmed_words)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import string
nltk.download('punkt')
nltk.download('stopwords')
text = "I am learning NLP and it is very exciting!!!"
text = text.lower()
text = text.translate(str.maketrans('', '', string.punctuation))
tokens = word_tokenize(text)
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in tokens if word not in stop_words]
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print(stemmed_words)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
txt="the movie was very good and very interesting"
token=word_tokenize(txt)
print(token)
stop_words=set(stopwords.words("english"))
filterwords=[]
for word in token:
    if word.lower() not in stop_words:
        filterwords.append(word)
print(filterwords)        
```

```python
from nltk.stem import WordNetLemmatizer
lemm=WordNetLemmatizer()
text="The children are playing with toys"
token=word_tokenize(text)
lemmas=[]
for word in token:
    lemmas.append(lemm.lemmatize(word))
print(lemmas)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
data = {
    "text": [
        "I love this product",
        "This is amazing",
        "Worst experience ever",
        "I hate this",
        "Very good service",
        "Not satisfied",
        "Excellent quality",
        "Bad product",
        "It is okay"
    ],

    "label": [
        "positive",
        "positive",
        "negative",
        "negative",
        "positive",
        "negative",
        "positive",
        "negative",
        "neutral"
    ]
}

df = pd.DataFrame(data)
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['label'], test_size=0.4, random_state=42
)
vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
model = MultinomialNB()
model.fit(X_train_vec, y_train)
y_pred = model.predict(X_test_vec)
print("Accuracy:", accuracy_score(y_test, y_pred))
sample = ["The phone is okay"]
sample_vec = vectorizer.transform(sample)
prediction = model.predict(sample_vec)
print(prediction[0])
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
data = {
    "text":[
        "I love this product",
        "This is amazing",
        "Worst experience ever",
        "I hate this",
        "Very good service",
        "Not satisfied",
        "Excellent quality",
        "Bad product",
        "It is okay",
        "Absolutely fantastic",
        "Very poor service",
        "I am happy",
        "Totally disappointed",
        "Great performance",
        "Terrible item",
        "Average experience",
        "Really enjoyed it",
        "Not good",
        "Best purchase ever",
        "Waste of money",
        "Nothing special",
        "Pretty decent"
    ],

    "label":[
        "positive",
        "positive",
        "negative",
        "negative",
        "positive",
        "negative",
        "positive",
        "negative",
        "neutral",
        "positive",
        "negative",
        "positive",
        "negative",
        "positive",
        "negative",
        "neutral",
        "positive",
        "negative",
        "positive",
        "negative",
        "neutral",
        "neutral"
    ]
}

df = pd.DataFrame(data)
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['label'], test_size=0.4, random_state=42
)
vectorizer = TfidfVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
model = MultinomialNB()
model.fit(X_train_vec, y_train)
y_pred = model.predict(X_test_vec)
print("Accuracy:", accuracy_score(y_test, y_pred))
sample = ["The movie is fantastic"]
sample_vec = vectorizer.transform(sample)
prediction = model.predict(sample_vec)
print("Prediction:", prediction[0])
```
