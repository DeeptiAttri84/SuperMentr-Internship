```python
import cv2
import numpy as np
image = cv2.imread("Sample.jpg")
print("Image Shape:", image.shape)
print("Pixel at (0,0):", image[0, 0])
print("Channels (BGR):", image[0, 0])
print("\nExplanation:")
print("Shape = (Height, Width, Channels)")
print("Pixel values range from 0 to 255")
print("Channels: B = Blue, G = Green, R = Red")
```
