```python
import bcrypt
import getpass
import re
import time

# In real applications, store this in a database
users_db = {}

MAX_ATTEMPTS = 3
LOCK_TIME = 30  # seconds


def is_strong_password(password):
    """Check password strength."""
    if len(password) < 8:
        return False, "Password must be at least 8 characters."

    if not re.search(r"[A-Z]", password):
        return False, "Add at least one uppercase letter."

    if not re.search(r"[a-z]", password):
        return False, "Add at least one lowercase letter."

    if not re.search(r"[0-9]", password):
        return False, "Add at least one number."

    if not re.search(r"[!@#$%^&*(),.?\":{}|<>]", password):
        return False, "Add at least one special character."

    return True, "Strong password."


def register():
    username = input("Enter username: ")

    if username in users_db:
        print("User already exists.")
        return

    password = getpass.getpass("Enter password: ")

    valid, message = is_strong_password(password)
    if not valid:
        print("Weak password:", message)
        return

    # Hash password
    hashed = bcrypt.hashpw(password.encode(), bcrypt.gensalt())

    users_db[username] = {
        "password": hashed,
        "attempts": 0,
        "lock_until": 0,
    }

    print("Registration successful.")


def login():
    username = input("Username: ")

    if username not in users_db:
        print("User not found.")
        return

    user = users_db[username]

    # Check lock
    if time.time() < user["lock_until"]:
        print("Account locked. Try later.")
        return

    password = getpass.getpass("Password: ")

    if bcrypt.checkpw(password.encode(), user["password"]):
        print("Login successful!")
        user["attempts"] = 0
    else:
        user["attempts"] += 1
        print("Incorrect password.")

        if user["attempts"] >= MAX_ATTEMPTS:
            user["lock_until"] = time.time() + LOCK_TIME
            user["attempts"] = 0
            print("Too many attempts. Account locked.")


def main():
    while True:
        print("\n1. Register\n2. Login\n3. Exit")
        choice = input("Choose option: ")

        if choice == "1":
            register()
        elif choice == "2":
            login()
        elif choice == "3":
            break
        else:
            print("Invalid option.")


if __name__ == "__main__":
    main()
```
