```python
import pandas as pd
data = {
    "Age": [25, 30, 22, 35, None],
    "Salary": [50000, 60000, 45000, 80000, 70000],
    "Experience": [2, 5, 1, 10, 7]
}
df = pd.DataFrame(data)
print("Top Rows:")
print(df.head())
max_values = df.max(numeric_only=True)
print("\nHighest Value in Each Column:")
print(max_values)
print("\nMissing Values:")
print(df.isnull().sum())
```
