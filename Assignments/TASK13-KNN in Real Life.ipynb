```python
"""
Introduction

K-Nearest Neighbors (KNN) is a simple machine learning algorithm that works based on similarity. Instead of learning complex patterns, it finds the most similar data points and uses them to make predictions or recommendations.

Many online platforms use this idea. For example, streaming services like Netflix recommend movies and shows by comparing your preferences with those of other users who have similar tastes.
"""
```

```python
"""
How Netflix-like Recommendation Works

A recommendation system using KNN follows a simple process:

The system collects ratings or watch history from users.

Each user’s preferences are stored as data (for example, ratings for different movies).

The algorithm compares users to find similar users.

It selects the K nearest neighbors, meaning the most similar users.

Movies or shows liked by those similar users are then recommended.

In simple words, if two people like many of the same movies, the system assumes they may enjoy similar content in the future.
"""
```

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np

# Movie rating data
data = np.array([
    [5,3,1],  # User A
    [4,3,1],  # User B
    [1,5,4]   # User C
])

# Train KNN model
model = NearestNeighbors(n_neighbors=2)
model.fit(data)

# Find neighbors of User A
distances, indices = model.kneighbors([data[0]])

print("Nearest Users:", indices)
print("Distances:", distances)
```
