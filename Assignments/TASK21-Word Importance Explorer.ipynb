```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love machine learning",
    "Machine learning is amazing",
    "I love coding",
    "Coding is fun",
    "Learning coding is useful"
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)

print("Feature Names:", vectorizer.get_feature_names_out())
print("TF-IDF Matrix:\n", X.toarray())
```

```python
"""TF (Term Frequency): How often a word appears in a document
IDF (Inverse Document Frequency): Penalizes common words
Words like machine, coding have high importance because:
They appear frequently in specific documents
But not in all documents"""
```
