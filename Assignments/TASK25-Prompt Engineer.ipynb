```python

def evaluate_prompt(prompt):
    score = 0
    
    if "ai/ml" in prompt.lower() or "python" in prompt.lower():
        score += 1   
    
    if "student" in prompt.lower() or "internship" in prompt.lower():
        score += 1  
    
    if "30-day" in prompt.lower() or "1-page" in prompt.lower() or "3" in prompt:
        score += 1  
    
    if "including" in prompt.lower():
        score += 1  
   
    if score >= 3:
        return "Strong Prompt → High quality output"
    else:
        return "Weak Prompt → Low quality output"


prompts = [
    "Write a resume",
    "Write a 1-page resume for a final-year AI/ML student with skills in Python, TensorFlow, NLP, and projects on stock prediction and book recommendation.",
    
    "Give business idea",
    "Suggest 3 AI-based startup ideas for students in India with low investment and high scalability.",
    
    "Make study plan",
    "Create a 30-day study plan for DSA and Machine Learning for internship preparation, including daily topics and practice questions."
]


for p in prompts:
    print("Prompt:", p)
    print("Evaluation:", evaluate_prompt(p))
    print("-" * 60)
```
