```python
students = {
    "Aman": 85,
    "Riya": 92,
    "Karan": 76,
    "Neha": 88,
    "Sahil": 69
}

topper = max(students, key=students.get)
print("Topper of the class:", topper)
print("Topper Marks:", students[topper])

total_marks = sum(students.values())
class_average = total_marks / len(students)
print("Class Average:",class_average)

print("\nStudent Grades:")
for name, marks in students.items():
    if marks >= 90:
        grade = "A+"
    elif marks >= 80:
        grade = "A"
    elif marks >= 70:
        grade = "B"
    elif marks >= 60:
        grade = "C"
    else:
        grade = "D"
    
    print(name, ":", marks, "-> Grade:", grade)
```
