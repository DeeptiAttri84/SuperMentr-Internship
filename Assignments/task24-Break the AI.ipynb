```python
def analyze_prompt(prompt):
    prompt = prompt.lower()
    if "bank" in prompt:
        return "Ambiguous: Could mean financial bank or river bank."

    elif "1 word" in prompt and "detail" in prompt:
        return "Contradiction: Cannot explain in 1 word and in detail simultaneously."
    
    elif "heavier" in prompt and "1 kg" in prompt:
        return "Both are equal (1 kg = 1 kg)."

    elif "better than another" in prompt:
        return "Bias detected: Cannot compare groups unfairly."

    elif not any(char.isalpha() for char in prompt):
        return "Invalid input: Cannot understand."

    else:
        return "Normal query: Can be answered."

test_prompts = [
    "Tell me about bank",
    "Explain in 1 word but in detail",
    "What is heavier: 1 kg iron or 1 kg cotton?",
    "Why is one group better than another?",
    "asdkjh123@#"
]

for p in test_prompts:
    print(f"Input: {p}")
    print("AI Response:", analyze_prompt(p))
    print("-" * 50)
```
