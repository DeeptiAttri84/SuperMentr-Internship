```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import string
```

```python
nltk.download("punkt")
nltk.download("stopwords")
```

```python
text="I absolutely loved this movie!"
text=text.lower()
print(text)
text=text.translate(str.maketrans('','',string.punctuation))
print(text)
```

```python
token=word_tokenize(text)
print(token)
stop_word=set(stopwords.words('english'))
filterwords=[word for word in token if word not in stop_word]
print(filterwords)
```

```python
stemmer=PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filterwords]
print(stemmed_words)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import string
nltk.download('punkt')
nltk.download('stopwords')
text = "I am learning NLP and it is very exciting!!!"
text = text.lower()
text = text.translate(str.maketrans('', '', string.punctuation))
tokens = word_tokenize(text)
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in tokens if word not in stop_words]
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print(stemmed_words)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
txt="the movie was very good and very interesting"
token=word_tokenize(txt)
print(token)
stop_words=set(stopwords.words("english"))
filterwords=[]
for word in token:
    if word.lower() not in stop_words:
        filterwords.append(word)
print(filterwords)        
```

```python
from nltk.stem import WordNetLemmatizer
lemm=WordNetLemmatizer()
text="The children are playing with toys"
token=word_tokenize(text)
lemmas=[]
for word in token:
    lemmas.append(lemm.lemmatize(word))
print(lemmas)
```
